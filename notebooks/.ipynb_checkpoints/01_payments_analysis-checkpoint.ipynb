```python
# === 01_payments_analysis.ipynb ===
import duckdb
import pandas as pd
from pathlib import Path

print("=== PAYMENTS DATA LAKE ANALYSIS ===\n")

con = duckdb.connect()

# Bronze Layer (Raw)
print("BRONZE LAYER (Raw Data)")
con.sql("""
    SELECT 
        COUNT(*) as total_records,
        COUNT(DISTINCT transaction_id) as unique_transactions,
        ROUND(AVG(amount), 2) as avg_amount,
        COUNT(CASE WHEN status = 'failed' THEN 1 END) as failed_count,
        MIN(timestamp) as earliest_tx,
        MAX(timestamp) as latest_tx
    FROM 'data/raw/date=*/payments_raw.parquet'
""").show()

# Silver Layer (Clean)
print("\nSILVER LAYER (Cleaned Data)")
con.sql("""
    SELECT 
        COUNT(*) as total_records,
        ROUND(AVG(amount), 2) as avg_amount,
        AVG(risk_score) as avg_risk_score,
        COUNT(CASE WHEN risk_score > 0 THEN 1 END) as high_risk_count,
        COUNT(DISTINCT user_id) as unique_users
    FROM 'data/silver/date=*/payments_silver.parquet'
""").show()

# Simple join example (Bronze + Silver insight)
print("\nHIGH-RISK TRANSACTIONS EXAMPLE")
con.sql("""
    SELECT 
        s.transaction_id,
        s.amount,
        s.risk_score,
        s.status
    FROM 'data/silver/date=*/payments_silver.parquet' s
    WHERE s.risk_score > 0
    LIMIT 5
""").show()
```
