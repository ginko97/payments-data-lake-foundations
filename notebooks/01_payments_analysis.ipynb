```python
import os
from pathlib import Path
import duckdb

# Fix working directory
os.chdir(Path.cwd().parent)
print("Working directory fixed to:", Path.cwd())

con = duckdb.connect()

print("=== BRONZE LAYER (Raw) ===")
con.sql("""
    SELECT 
        COUNT(*) as total_records,
        ROUND(AVG(amount), 2) as avg_amount,
        COUNT(CASE WHEN status = 'failed' THEN 1 END) as failed_tx
    FROM 'data/raw/date=*/payments_raw.parquet'
""").show()

print("\n=== SILVER LAYER (Clean) ===")
con.sql("""
    SELECT 
        COUNT(*) as total_records,
        ROUND(AVG(risk_score), 2) as avg_risk_score,
        COUNT(CASE WHEN risk_score > 0 THEN 1 END) as high_risk
    FROM 'data/silver/date=*/payments_silver.parquet'
""").show()
```
